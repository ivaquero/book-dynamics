```python
import cvxpy as cp
import numpy as np
import pandas as pd
```

## 1.2

### 1.1

```python
x = cp.Variable((2), nonneg=True, integer=True)
objective = cp.Maximize(x[0] * 4000 + x[1] * 3000)

constraints = [x[0] * 2 + x[1] <= 10, x[0] + x[1] <= 8, x[1] <= 7]

problem = cp.Problem(objective, constraints)
problem.solve()

print(f"Optimal value: {problem.value}")
print(f"x = {x.value}")
```

### 1.3

```python
x1, x2, x3 = (
    cp.Variable(nonneg=True),
    cp.Variable(nonneg=True),
    cp.Variable(nonneg=True),
)

objective = cp.Maximize(x1 * 2 + x2 * 3 - x3 * 5)

constraints = [x1 + x2 + x3 == 7, x1 * 2 - x2 * 5 + x3 >= 10, x1 + x2 * 3 + x3 <= 12]

problem = cp.Problem(objective, constraints)
problem.solve()

print(f"Optimal value: {problem.value:.3f}")
print(f"x1 = {x1.value:.3f}\nx2 = {x2.value:.3f}\nx3 = {x3.value:.3f}")
```

### 1.4

```python
x = cp.Variable((4, 4), nonneg=True, integer=True)

objective = cp.Minimize(
    2800 * sum(x[0:4, 0])
    + 4500 * sum(x[0:3, 1])
    + 6000 * sum(x[0:2, 2])
    + 7300 * x[0, 3]
)

constraints = [
    sum(x[0, :]) >= 15,
    sum(x[0, 1:]) + sum(x[1, 0:4]) >= 10,
    sum(x[0, 2:]) + sum(x[1, 1:3]) + sum(x[2, 0:2]) >= 20,
    x[0, 3] + x[1, 2] + x[2, 1] + x[3, 0] >= 12,
    x[1, 3] == 0,
    x[2, 2] == 0,
    x[2, 3] == 0,
    x[3, 1] == 0,
    x[3, 2] == 0,
    x[3, 3] == 0,
]

problem = cp.Problem(objective, constraints)
problem.solve()

print(f"Optimal value: {problem.value:.0f}")
print(f"x = {np.abs(x.value)}")
```
